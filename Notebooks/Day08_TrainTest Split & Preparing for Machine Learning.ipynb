```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("C:\Customer_Churn_Analysis\Data\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df_ml = df.drop(columns=["customerID"])

print(df_ml.shape)
```

```python
X = df_ml.drop(columns=["Churn"])
y = df_ml["Churn"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print("Original shape:", X.shape)
print("Encoded shape:", X_encoded.shape)
```

```python
print("X shape:", X_encoded.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training set:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting set:")
print(y_test.value_counts(normalize=True) * 100)
```

### Train/Test Split

- The dataset was divided into training and testing sets.
- 80% of the data was used for training and 20% for testing.
- The training set contains 5,634 customers.
- The testing set contains 1,409 customers.
- `stratify=y` was used to preserve the original Churn distribution.
- The Churn distribution remained almost identical in both sets.
- The training set contains approximately 73.46% No Churn and 26.54% Churn.
- The testing set contains approximately 73.46% No Churn and 26.54% Churn.
