```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("C:\Customer_Churn_Analysis\Data\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df_ml = df.drop(columns=["customerID"])

X = df_ml.drop(columns=["Churn"])
y = df_ml["Churn"]
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)

model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
y_pred = model.predict(X_test_scaled)

print(y_pred[:20])
```

```python
print("Number of predictions:", len(y_pred))
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_test, y_pred, pos_label="Yes")
recall = recall_score(y_test, y_pred, pos_label="Yes")
f1 = f1_score(y_test, y_pred, pos_label="Yes")

print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

## Day 9 Findings

- The problem is a binary classification problem because the target variable has two classes: No and Yes.
- Logistic Regression was used as a baseline classification model.
- The model was trained on the training data and evaluated on unseen test data.
- The model achieved an accuracy of approximately 80.62%.
- The confusion matrix showed 925 true negatives, 110 false positives, 163 false negatives, and 211 true positives.
- Precision for the Yes class was approximately 65.73%.
- Recall for the Yes class was approximately 56.42%.
- F1-score was approximately 60.72%.
- The model correctly identified 211 churned customers but missed 163 actual churned customers.
- `predict()` returns the final predicted class, while `predict_proba()` returns the probability for each class.
- The class order was confirmed using `model.classes_`, which returned ['No', 'Yes'].
- The default classification threshold is approximately 0.5.
- Lowering the threshold may identify more potential churners, but it can also increase false positives.
