```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:\Customer_Churn_Analysis\Data\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# Check the data type and first values of TotalCharges
print(df["TotalCharges"].dtype)
print(df["TotalCharges"].head(10))
```

```python
# Convert TotalCharges to numeric temporarily
total_charges_numeric = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Count values that could not be converted to numbers
print("Non-numeric values:", total_charges_numeric.isna().sum())
```

```python
# Find the rows where TotalCharges could not be converted to a number
invalid_total_charges = df[
    pd.to_numeric(df["TotalCharges"], errors="coerce").isna()
]

# Display the relevant columns
print(invalid_total_charges[["customerID", "tenure", "TotalCharges", "Churn"]])
```

```python
# Replace blank spaces with NaN
df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)

# Convert TotalCharges from object to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])

# Check the new data type
print(df["TotalCharges"].dtype)
```

```python
# Check for missing values in TotalCharges
print("Missing TotalCharges:", df["TotalCharges"].isna().sum())
```

```python
# Display customers with missing TotalCharges
missing_total_charges = df[df["TotalCharges"].isna()]

print(missing_total_charges[["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]])
```

```python
# Check the tenure values for all customers with missing TotalCharges
print(
    df.loc[df["TotalCharges"].isna(), "tenure"].value_counts()
)
```

```python
# Fill missing TotalCharges with 0 for customers with tenure = 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)

# Check if any missing values remain
print("Missing TotalCharges:", df["TotalCharges"].isna().sum())
```

```python
# Check the data types after cleaning
print(df.dtypes)
```

```python
# Check the range of important numerical columns
print("Tenure:")
print("Min:", df["tenure"].min())
print("Max:", df["tenure"].max())

print("\nMonthly Charges:")
print("Min:", df["MonthlyCharges"].min())
print("Max:", df["MonthlyCharges"].max())

print("\nTotal Charges:")
print("Min:", df["TotalCharges"].min())
print("Max:", df["TotalCharges"].max())
```

```python

# Check for missing values in all columns
print("Missing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

- The `TotalCharges` column was originally stored as an object data type.
