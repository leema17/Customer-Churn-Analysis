```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:\Customer_Churn_Analysis\Data\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# Calculate churn rate by partner status
partner_churn = pd.crosstab(
    df["Partner"],
    df["Churn"],
    normalize="index"
) * 100

print(partner_churn)
```

```python
# Calculate churn rate by dependent status
dependents_churn = pd.crosstab(
    df["Dependents"],
    df["Churn"],
    normalize="index"
) * 100

print(dependents_churn)
```

```python
# Calculate churn rate by internet service type
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

print(internet_churn)
```

```python
# Calculate churn rate by online security service
security_churn = pd.crosstab(
    df["OnlineSecurity"],
    df["Churn"],
    normalize="index"
) * 100

print(security_churn)
```

```python
# Calculate churn rate by technical support service
techsupport_churn = pd.crosstab(
    df["TechSupport"],
    df["Churn"],
    normalize="index"
) * 100

print(techsupport_churn)
```

```python
# Calculate churn rate by payment method
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

print(payment_churn)
```

## Day 4 Findings

### Partner and Churn
- Customers without a partner have a churn rate of 32.96%.
- Customers with a partner have a churn rate of 19.66%.
- Partner status shows a noticeable association with churn.

### Dependents and Churn
- Customers without dependents have a churn rate of 31.28%.
- Customers with dependents have a churn rate of 15.45%.
- Dependents status shows a noticeable association with churn.

### Internet Service and Churn
- Fiber optic customers have the highest churn rate at 41.89%.
- DSL customers have a churn rate of 18.96%.
- Customers without internet service have the lowest churn rate at 7.40%.
- Internet service type appears to be strongly associated with churn.

### Online Security and Churn
- Customers without Online Security have a churn rate of 41.77%.
- Customers with Online Security have a churn rate of 14.61%.
- Customers without internet service have a churn rate of 7.40%.
- Online Security shows a strong association with churn.

### Tech Support and Churn
- Customers without Tech Support have a churn rate of 41.64%.
- Customers with Tech Support have a churn rate of 15.17%.
- Customers without internet service have a churn rate of 7.40%.
- Tech Support shows a strong association with churn.

### Payment Method and Churn
- Customers using Electronic check have the highest churn rate at 45.29%.
- Customers using Credit card (automatic) have the lowest churn rate at 15.24%.
- Payment method shows a strong association with churn.

### Overall Day 4 Insight
- Several customer service and account characteristics show noticeable differences in churn rates.
- The strongest differences observed today were related to Internet Service, Online Security, Tech Support, and Payment Method.
- These findings show associations in the dataset and do not prove that any individual factor directly causes customer churn.
